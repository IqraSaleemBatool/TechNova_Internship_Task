##Now Creating Our 5 KPIs

```python
# Calculate KPIs

total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
total_cogs = df['COGS'].sum()
total_units = df['Units Sold'].sum()

profit_margin = (total_profit / total_sales) * 100

print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Total COGS: ${total_cogs:,.2f}")
print(f"Profit Margin: {profit_margin:.2f}%")
print(f"Total Units Sold: {total_units:,.0f}")
```

```python
import dash
from dash import dcc, html, Input, Output
import plotly.express as px
```

Prepare dashboard data

```python
dashboard_df = df.copy()

dashboard_df['Year'] = dashboard_df['Date'].dt.year

dashboard_df['Month-Year'] = dashboard_df['Date'].dt.strftime('%Y-%m')
```

Checking dataset

```python
dashboard_df[['Date', 'Year', 'Month-Year']].head()
```

Create the Dash app

```python
app = dash.Dash(__name__)

app.layout = html.Div([

    html.H1(
        "Financial KPI Dashboard",
        style={
            'textAlign': 'center',
            'marginBottom': '25px'
        }
    ),

    html.Div([

        html.Div([
            html.Label("Country"),
            dcc.Dropdown(
                id='country-filter',
                options=[
                    {'label': 'All Countries', 'value': 'All'}
                ] + [
                    {'label': country, 'value': country}
                    for country in sorted(dashboard_df['Country'].unique())
                ],
                value='All',
                clearable=False
            )
        ], style={'width': '23%', 'display': 'inline-block', 'marginRight': '2%'}),

        html.Div([
            html.Label("Segment"),
            dcc.Dropdown(
                id='segment-filter',
                options=[
                    {'label': 'All Segments', 'value': 'All'}
                ] + [
                    {'label': segment, 'value': segment}
                    for segment in sorted(dashboard_df['Segment'].unique())
                ],
                value='All',
                clearable=False
            )
        ], style={'width': '23%', 'display': 'inline-block', 'marginRight': '2%'}),

        html.Div([
            html.Label("Product"),
            dcc.Dropdown(
                id='product-filter',
                options=[
                    {'label': 'All Products', 'value': 'All'}
                ] + [
                    {'label': product, 'value': product}
                    for product in sorted(dashboard_df['Product'].unique())
                ],
                value='All',
                clearable=False
            )
        ], style={'width': '23%', 'display': 'inline-block', 'marginRight': '2%'}),

        html.Div([
            html.Label("Year"),
            dcc.Dropdown(
                id='year-filter',
                options=[
                    {'label': 'All Years', 'value': 'All'}
                ] + [
                    {'label': str(year), 'value': year}
                    for year in sorted(dashboard_df['Year'].unique())
                ],
                value='All',
                clearable=False
            )
        ], style={'width': '23%', 'display': 'inline-block'})

    ], style={'marginBottom': '30px'}),

    html.Div([

        html.Div([
            html.H4("Total Sales"),
            html.H2(id='total-sales')
        ], style={'width': '18%', 'display': 'inline-block', 'textAlign': 'center'}),

        html.Div([
            html.H4("Total Profit"),
            html.H2(id='total-profit')
        ], style={'width': '18%', 'display': 'inline-block', 'textAlign': 'center'}),

        html.Div([
            html.H4("Total COGS"),
            html.H2(id='total-cogs')
        ], style={'width': '18%', 'display': 'inline-block', 'textAlign': 'center'}),

        html.Div([
            html.H4("Profit Margin"),
            html.H2(id='profit-margin')
        ], style={'width': '18%', 'display': 'inline-block', 'textAlign': 'center'}),

        html.Div([
            html.H4("Units Sold"),
            html.H2(id='total-units')
        ], style={'width': '18%', 'display': 'inline-block', 'textAlign': 'center'})

    ], style={'marginBottom': '35px'}),

    html.Div([
        dcc.Graph(id='monthly-trend')
    ]),

    html.Div([
        html.Div([
            dcc.Graph(id='sales-segment')
        ], style={'width': '48%', 'display': 'inline-block'}),

        html.Div([
            dcc.Graph(id='profit-country')
        ], style={'width': '48%', 'display': 'inline-block'})
    ]),

    html.Div([
        dcc.Graph(id='profit-product')
    ])

])
```

Add the functionality

```python
@app.callback(
    [
        Output('total-sales', 'children'),
        Output('total-profit', 'children'),
        Output('total-cogs', 'children'),
        Output('profit-margin', 'children'),
        Output('total-units', 'children'),
        Output('monthly-trend', 'figure'),
        Output('sales-segment', 'figure'),
        Output('profit-country', 'figure'),
        Output('profit-product', 'figure')
    ],
    [
        Input('country-filter', 'value'),
        Input('segment-filter', 'value'),
        Input('product-filter', 'value'),
        Input('year-filter', 'value')
    ]
)
def update_dashboard(country, segment, product, year):

    filtered = dashboard_df.copy()

    if country != 'All':
        filtered = filtered[filtered['Country'] == country]

    if segment != 'All':
        filtered = filtered[filtered['Segment'] == segment]

    if product != 'All':
        filtered = filtered[filtered['Product'] == product]

    if year != 'All':
        filtered = filtered[filtered['Year'] == year]

    total_sales = filtered['Sales'].sum()
    total_profit = filtered['Profit'].sum()
    total_cogs = filtered['COGS'].sum()
    total_units = filtered['Units Sold'].sum()

    profit_margin = (
        total_profit / total_sales * 100
        if total_sales != 0 else 0
    )

    monthly = (
        filtered.groupby('Date', as_index=False)
        .agg({
            'Sales': 'sum',
            'COGS': 'sum',
            'Profit': 'sum'
        })
        .sort_values('Date')
    )

    sales_segment = (
        filtered.groupby('Segment', as_index=False)['Sales']
        .sum()
        .sort_values('Sales', ascending=False)
    )

    profit_country = (
        filtered.groupby('Country', as_index=False)['Profit']
        .sum()
        .sort_values('Profit', ascending=False)
    )

    profit_product = (
        filtered.groupby('Product', as_index=False)['Profit']
        .sum()
        .sort_values('Profit', ascending=False)
    )

    fig_monthly = px.line(
        monthly,
        x='Date',
        y=['Sales', 'COGS', 'Profit'],
        markers=True,
        title='Monthly Financial Performance'
    )

    fig_segment = px.bar(
        sales_segment,
        x='Segment',
        y='Sales',
        title='Sales by Customer Segment',
        text_auto='.2s'
    )

    fig_country = px.bar(
        profit_country,
        x='Country',
        y='Profit',
        title='Profit by Country',
        text_auto='.2s'
    )

    fig_product = px.bar(
        profit_product,
        x='Product',
        y='Profit',
        title='Profit by Product',
        text_auto='.2s'
    )

    return (
        f"${total_sales:,.2f}",
        f"${total_profit:,.2f}",
        f"${total_cogs:,.2f}",
        f"{profit_margin:.2f}%",
        f"{total_units:,.0f}",
        fig_monthly,
        fig_segment,
        fig_country,
        fig_product
    )
```
