##Now Creating Our 5 KPIs

```python
# Calculate KPIs

total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
total_cogs = df['COGS'].sum()
total_units = df['Units Sold'].sum()

profit_margin = (total_profit / total_sales) * 100

print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Total COGS: ${total_cogs:,.2f}")
print(f"Profit Margin: {profit_margin:.2f}%")
print(f"Total Units Sold: {total_units:,.0f}")
```

```python
%%writefile dashboard/app.py

import pandas as pd
import plotly.express as px
from dash import Dash, dcc, html, Input, Output


#  Load and Prepare Data

df = pd.read_excel("Financial Sample.xlsx")

# Clean column names
df.columns = df.columns.str.strip()

# Convert Date column
df["Date"] = pd.to_datetime(df["Date"])

# Create Year and Month-Year columns
df["Year"] = df["Date"].dt.year
df["Month-Year"] = df["Date"].dt.strftime("%Y-%m")

# Dashboard dataframe
dashboard_df = df.copy()


#  Create Dashboard APP


app = Dash(__name__)

app.title = "Financial KPI Dashboard"



#  Filter options


countries = sorted(
    dashboard_df["Country"].dropna().unique()
)

segments = sorted(
    dashboard_df["Segment"].dropna().unique()
)

products = sorted(
    dashboard_df["Product"].dropna().unique()
)

years = sorted(
    dashboard_df["Year"].dropna().unique()
)


#  Dashboard Layout

app.layout = html.Div([


    # Dashboard Title


    html.H1(
        "Financial KPI Dashboard",
        style={
            "textAlign": "center",
            "marginBottom": "25px"
        }
    ),



    # Filters

    html.Div([

        # Country
        html.Div([
            html.Label("Country"),

            dcc.Dropdown(
                id="country-filter",

                options=[
                    {
                        "label": "All Countries",
                        "value": "All"
                    }
                ] + [
                    {
                        "label": country,
                        "value": country
                    }
                    for country in countries
                ],

                value="All",
                clearable=False
            )

        ], style={
            "width": "23%",
            "display": "inline-block",
            "marginRight": "2%"
        }),


        # Segment
        html.Div([
            html.Label("Segment"),

            dcc.Dropdown(
                id="segment-filter",

                options=[
                    {
                        "label": "All Segments",
                        "value": "All"
                    }
                ] + [
                    {
                        "label": segment,
                        "value": segment
                    }
                    for segment in segments
                ],

                value="All",
                clearable=False
            )

        ], style={
            "width": "23%",
            "display": "inline-block",
            "marginRight": "2%"
        }),


        # Product
        html.Div([
            html.Label("Product"),

            dcc.Dropdown(
                id="product-filter",

                options=[
                    {
                        "label": "All Products",
                        "value": "All"
                    }
                ] + [
                    {
                        "label": product,
                        "value": product
                    }
                    for product in products
                ],

                value="All",
                clearable=False
            )

        ], style={
            "width": "23%",
            "display": "inline-block",
            "marginRight": "2%"
        }),


        # Year

        html.Div([
            html.Label("Year"),

            dcc.Dropdown(
                id="year-filter",

                options=[
                    {
                        "label": "All Years",
                        "value": "All"
                    }
                ] + [
                    {
                        "label": str(year),
                        "value": year
                    }
                    for year in years
                ],

                value="All",
                clearable=False
            )

        ], style={
            "width": "23%",
            "display": "inline-block"
        })

    ], style={
        "marginBottom": "30px"
    }),



    # 5. KPI Cards

    html.Div([

        # Total Sales
        html.Div([
            html.H4("Total Sales"),
            html.H2(id="total-sales")
        ], style={
            "width": "18%",
            "display": "inline-block",
            "textAlign": "center"
        }),


        # Total Profit
        html.Div([
            html.H4("Total Profit"),
            html.H2(id="total-profit")
        ], style={
            "width": "18%",
            "display": "inline-block",
            "textAlign": "center"
        }),


        # Total COGS
        html.Div([
            html.H4("Total COGS"),
            html.H2(id="total-cogs")
        ], style={
            "width": "18%",
            "display": "inline-block",
            "textAlign": "center"
        }),


        # Profit Margin
        html.Div([
            html.H4("Profit Margin"),
            html.H2(id="profit-margin")
        ], style={
            "width": "18%",
            "display": "inline-block",
            "textAlign": "center"
        }),


        # Units Sold
        html.Div([
            html.H4("Units Sold"),
            html.H2(id="total-units")
        ], style={
            "width": "18%",
            "display": "inline-block",
            "textAlign": "center"
        })

    ], style={
        "marginBottom": "35px"
    }),



    #  Monthly Financial Performance

    html.Div([
        dcc.Graph(
            id="monthly-trend"
        )
    ]),



    #  Sales by  Segment + Profit by country


    html.Div([

        html.Div([
            dcc.Graph(
                id="sales-segment"
            )
        ], style={
            "width": "48%",
            "display": "inline-block"
        }),


        html.Div([
            dcc.Graph(
                id="profit-country"
            )
        ], style={
            "width": "48%",
            "display": "inline-block"
        })

    ]),


    #  PROFIT BY PRODUCT

    html.Div([
        dcc.Graph(
            id="profit-product"
        )
    ])

], style={
    "padding": "30px",
    "fontFamily": "Arial"
})


#  Dashboard callback

@app.callback(

    Output("total-sales", "children"),
    Output("total-profit", "children"),
    Output("total-cogs", "children"),
    Output("profit-margin", "children"),
    Output("total-units", "children"),

    Output("monthly-trend", "figure"),
    Output("sales-segment", "figure"),
    Output("profit-country", "figure"),
    Output("profit-product", "figure"),

    Input("country-filter", "value"),
    Input("segment-filter", "value"),
    Input("product-filter", "value"),
    Input("year-filter", "value")
)


def update_dashboard(
    country,
    segment,
    product,
    year
):

    # Start with complete dataset


    filtered = dashboard_df.copy()


    # Country Filter


    if country != "All":

        filtered = filtered[
            filtered["Country"] == country
        ]


    # Segment Filter

    if segment != "All":

        filtered = filtered[
            filtered["Segment"] == segment
        ]


    # Product Filter

    if product != "All":

        filtered = filtered[
            filtered["Product"] == product
        ]


    # Year Filter

    if year != "All":

        filtered = filtered[
            filtered["Year"] == year
        ]



    #  KPI Calculations

    total_sales = filtered["Sales"].sum()

    total_profit = filtered["Profit"].sum()

    total_cogs = filtered["COGS"].sum()

    total_units = filtered["Units Sold"].sum()


    # Profit Margin

    profit_margin = (

        total_profit / total_sales * 100

        if total_sales != 0

        else 0
    )


    #  Monthly financial performance

    monthly = (

        filtered
        .groupby("Date", as_index=False)
        .agg({
            "Sales": "sum",
            "COGS": "sum",
            "Profit": "sum"
        })
        .sort_values("Date")
    )


    fig_monthly = px.line(

        monthly,

        x="Date",

        y=[
            "Sales",
            "COGS",
            "Profit"
        ],

        markers=True,

        title="Monthly Financial Performance"
    )


    fig_monthly.update_layout(
        template="plotly_white",
        xaxis_title="Date",
        yaxis_title="Amount"
    )



    #  Sales by customer segment

    sales_segment = (

        filtered
        .groupby(
            "Segment",
            as_index=False
        )["Sales"]
        .sum()
        .sort_values(
            "Sales",
            ascending=False
        )
    )


    fig_segment = px.bar(

        sales_segment,

        x="Segment",

        y="Sales",

        title="Sales by Customer Segment",

        text_auto=".2s"
    )


    fig_segment.update_layout(
        template="plotly_white",
        xaxis_title="Customer Segment",
        yaxis_title="Sales"
    )


    #  Profit by country


    profit_country = (

        filtered
        .groupby(
            "Country",
            as_index=False
        )["Profit"]
        .sum()
        .sort_values(
            "Profit",
            ascending=False
        )
    )


    fig_country = px.bar(

        profit_country,

        x="Country",

        y="Profit",

        title="Profit by Country",

        text_auto=".2s"
    )


    fig_country.update_layout(
        template="plotly_white",
        xaxis_title="Country",
        yaxis_title="Profit"
    )


    #  Profit by product


    profit_product = (

        filtered
        .groupby(
            "Product",
            as_index=False
        )["Profit"]
        .sum()
        .sort_values(
            "Profit",
            ascending=False
        )
    )


    fig_product = px.bar(

        profit_product,

        x="Product",

        y="Profit",

        title="Profit by Product",

        text_auto=".2s"
    )


    fig_product.update_layout(
        template="plotly_white",
        xaxis_title="Product",
        yaxis_title="Profit"
    )



    #  Return Results

    return (

        f"${total_sales:,.2f}",

        f"${total_profit:,.2f}",

        f"${total_cogs:,.2f}",

        f"{profit_margin:.2f}%",

        f"{total_units:,.0f}",

        fig_monthly,

        fig_segment,

        fig_country,

        fig_product
    )



#  Run Dash APP


if __name__ == "__main__":

    app.run(
        debug=False,
        jupyter_mode="inline"
    )
```
